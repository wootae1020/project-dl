# 황사

```python
os.chdir('./data/황사')
os.getcwd()
```

```python
황사리스트 = os.listdir()
황사리스트 = list(set(황사리스트)-set([file for file in 황사리스트 if ('SNDE' in file) or ('WSPF' in file)]))
```

```python
df.columns
```

```python
use_cols = ['일자', '지점명', '황사관측']
df = pd.read_csv(황사리스트[0], encoding='CP949')
df.columns = ['일자', '지점번호', '지점명', '황사관측', '평균 미세먼지 농도(㎍/㎥)', '평균 풍속(m/s)',
       '최대 풍속(m/s)', '최대 풍속 풍향(16방위)', '최다풍향(16방위)', '기사', '최단시정(10m)',
       '대기확산지수:+03시간', '+06시간', '+09시간', '+12시간', '+15시간', '+18시간', '+21시간',
       '+24시간']
df = df[use_cols]
for file_path1 in 황사리스트[1:]:
    df2 = pd.read_csv(file_path1, encoding='CP949')
    df2.columns = ['일자', '지점번호', '지점명', '황사관측', '평균 미세먼지 농도(㎍/㎥)', '평균 풍속(m/s)',
       '최대 풍속(m/s)', '최대 풍속 풍향(16방위)', '최다풍향(16방위)', '기사', '최단시정(10m)',
       '대기확산지수:+03시간', '+06시간', '+09시간', '+12시간', '+15시간', '+18시간', '+21시간',
       '+24시간']
    df2 =df2[use_cols]
    if df.columns.equals(df2.columns)==True: 
        df = pd.concat([df,df2], axis=0)
        #print(df.shape)
    else : 
        print('실패')
        print(df.columns)
        print(df2.columns)
```

```python
df2 = df[df['지점명'].str.contains('서울')]
```

```python
df2.sort_values(by='일자').drop_duplicates().reset_index(drop=True).to_csv('황사.csv', index=False)
```
